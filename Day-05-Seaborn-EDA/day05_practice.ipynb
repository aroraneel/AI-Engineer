```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# TASK 1 — Load the data & first look

# 1a. Read "employees.csv" into a DataFrame called df
df = pd.read_csv(r"A:\Ai-Engineer\Day-05-Seaborn-EDA\employees.csv")

# 1b. Print df.head()
print(df.head())

# 1c. Print df.shape
print(df.shape)

# 1d. Print df.info()
print(df.info())
```

```python
# TASK 2 — Check missing values & summary statistics

# 2a. Print df.isnull().sum() — this shows how many missing values are in
#     each column
print(df.isnull().sum())

# 2b. Print df.describe() — quick statistics for numeric columns
print(df.describe())
```

```python
# TASK 3 — Distribution of a numeric column

# 3a. Create a histogram of the "Exam_Score" column using sns.histplot,
#     with bins=15
sns.histplot(data=df, x="Exam_Score", bins=15)

# 3b. Add title "Exam Score Distribution"
plt.title("Exam Score Distribution")

# 3c. Show the plot
plt.show()
```

```python
# TASK 4 — Spotting outliers with a box plot

# 4a. Create a box plot showing "Salary" (y-axis) broken down by
#     "Department" (x-axis)
sns.boxplot(data=df, x="Department", y="Salary")

# 4b. Add title "Salary by Department"
plt.title("Salary by Department")

# 4c. Show the plot
plt.show()
```

```python
# TASK 5 — Relationships (scatter plot with hue)

# 5a. Create a scatter plot: x="Study_Hours", y="Exam_Score", colored by
#     "Gender" (hue)
sns.scatterplot(data=df, x="Study_Hours", y="Exam_Score", hue="Gender")

# 5b. Add title "Study Hours vs Exam Score by Gender"
plt.title("Study Hours vs Exam Score by Gender")

# 5c. Show the plot
plt.show()
```

```python
# TASK 6 — Category counts

# 6a. Create a count plot showing how many employees are in each
#     "Department"
sns.countplot(data=df, x="Department")

# 6b. Add title "Employee Count by Department"
plt.title("Employee Count by Department")

# 6c. Show the plot
plt.show()
```

```python
# TASK 7 — Correlation heatmap

# 7a. Compute the correlation matrix of the numeric columns in df
#     (hint: df.corr(numeric_only=True))
corr_matrix = df.corr(numeric_only=True)

# 7b. Create a heatmap of that correlation matrix, with annot=True
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")

# 7c. Add title "Correlation Heatmap"
plt.title("Correlation Heatmap")

# 7d. Show the plot
plt.show()
```

```python
# TASK 8 — Full picture with a pair plot

# 8a. Create a pairplot of the entire df (just sns.pairplot(df) is fine —
#     it will auto-select numeric columns)
sns.pairplot(df)

# 8b. Show the plot
plt.show()

# 8c. Insight: Study_Hours has a strong positive correlation with Exam_Score
# (0.88), while Salary appears to depend much more on Department than on
# any individual factor like age, study hours, or exam performance.
```
